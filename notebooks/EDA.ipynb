# Exploratory Data Analysis: Travel Recommendation & Multi-Objective Planning
This notebook explores the dataset generated for South India tourism POIs, synthetic user latent profiles, and interaction ratings.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

places = pd.read_csv('../data/processed/places.csv')
users = pd.read_csv('../data/processed/users.csv')
ratings = pd.read_csv('../data/processed/ratings.csv')

print(f"Places shape: {places.shape}")
print(f"Users shape: {users.shape}")
print(f"Ratings shape: {ratings.shape}")
```

### 1. Places Category Distribution & Cost Analysis

```python
plt.figure(figsize=(12, 4))
plt.subplot(1, 2, 1)
places['category'].value_counts().plot(kind='bar', color='teal')
plt.title('Places per Category')
plt.xticks(rotation=45)

plt.subplot(1, 2, 2)
sns.boxplot(x='category', y='avg_cost_inr', data=places)
plt.title('Cost Distribution by Category')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### 2. Rating Distribution and User Interaction Sparsity

```python
plt.figure(figsize=(10, 4))
plt.subplot(1, 2, 1)
ratings['rating'].value_counts().sort_index().plot(kind='bar', color='coral')
plt.title('Rating Distribution (1 to 5 Stars)')

plt.subplot(1, 2, 2)
ratings_per_user = ratings.groupby('user_id').size()
sns.histplot(ratings_per_user, bins=25, kde=True, color='purple')
plt.title('Reviews per User Distribution')
plt.tight_layout()
plt.show()
```
